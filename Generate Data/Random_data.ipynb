```python
# Generate Gaussian Data for 2 Classes

import numpy as np

np.random.seed(0)

n = 400
half = n // 2

# Class 0 centered at (-2, -2)
class0 = np.random.randn(half, 2) + np.array([-2, -2])

# Class 1 centered at (2, 2)
class1 = np.random.randn(half, 2) + np.array([2, 2])

X = np.vstack((class0, class1))
y = np.vstack((np.zeros((half,1)), np.ones((half,1))))
print("X shape:", X.shape) # (400, 2)
print("y shape:", y.shape) # (400, 1)
print("First 5 samples:\n", X[:5])
print("First 5 labels:\n", y[:5])
```

```python
# Generate Gaussian Data with Different Covariances

mean0 = [-2, -2]
cov0 = [[1, 0.8], [0.8, 1]]

mean1 = [2, 2]
cov1 = [[1, -0.6], [-0.6, 1]]

class0 = np.random.multivariate_normal(mean0, cov0, 200)
class1 = np.random.multivariate_normal(mean1, cov1, 200)

X = np.vstack((class0, class1))
y = np.vstack((np.zeros((200,1)), np.ones((200,1))))

print("X shape:", X.shape) # (400, 2)
print("y shape:", y.shape) # (400, 1)
print("First 5 samples:\n", X[:5])
print("First 5 labels:\n", y[:5])
```

```python
# Generate Uniform Data for 2 Classes

class0 = np.random.uniform(-4, -1, (200,2))
class1 = np.random.uniform(1, 4, (200,2))

X = np.vstack((class0, class1))
y = np.vstack((np.zeros((200,1)), np.ones((200,1))))
print("X shape:", X.shape) # (400, 2)
print("y shape:", y.shape) # (400, 1)   
print("First 5 samples:\n", X[:5])
print("First 5 labels:\n", y[:5])
```

```python
# Generate Circular Data for 2 Classes

theta = np.linspace(0, 2*np.pi, 200)

r1 = 1
r2 = 3

class0 = np.c_[r1*np.cos(theta), r1*np.sin(theta)]
class1 = np.c_[r2*np.cos(theta), r2*np.sin(theta)]

X = np.vstack((class0, class1))
y = np.vstack((np.zeros((200,1)), np.ones((200,1))))

print("X shape:", X.shape) # (400, 2)
print("y shape:", y.shape) # (400, 1)   
print("First 5 samples:\n", X[:5])
print("First 5 labels:\n", y[:5])
```

```python
# Generate XOR Data for 2 Classes

X = np.random.randn(400,2)
y = np.logical_xor(X[:,0] > 0, X[:,1] > 0).astype(int)
y = y.reshape(-1,1)
print("X shape:", X.shape) # (400, 2)
print("y shape:", y.shape) # (400, 1)   
print("First 5 samples:\n", X[:5])
print("First 5 labels:\n", y[:5])
```

```python
# Imbalanced Data: 350 samples of Class 0 and 50 samples of Class 1

class0 = np.random.randn(350,2) + [-2,-2]
class1 = np.random.randn(50,2) + [2,2]

X = np.vstack((class0, class1))
y = np.vstack((np.zeros((350,1)), np.ones((50,1))))
print("X shape:", X.shape) # (400, 2)
print("y shape:", y.shape) # (400, 1)   
print("First 5 samples:\n", X[:5])
print("First 5 labels:\n", y[:5])
```

```python
# Using Random Linear Boundary (General Method)

np.random.seed(0)

X = np.random.randn(400,2)

real_w = np.array([[1], [-2]])
bias = 0.5

z = X @ real_w + bias
y = (z > 0).astype(int)

print(z[:5])  # Print first 5 linear combinations
print(y[:5])  # Print first 5 labels
```

```python
# Add Noise: Flip 10% of the labels randomly

noise_idx = np.random.choice(400, 40)  # flip 10% labels
y[noise_idx] = 1 - y[noise_idx]
print("First 5 labels:\n", y[:5])
```

```python
# Spiral Data for 2 Classes
n = 200
theta = np.sqrt(np.random.rand(n))*2*np.pi

r_a = 2*theta + np.pi
data_a = np.c_[np.cos(theta)*r_a, np.sin(theta)*r_a]

r_b = -2*theta - np.pi
data_b = np.c_[np.cos(theta)*r_b, np.sin(theta)*r_b]

X = np.vstack((data_a, data_b))
y = np.vstack((np.zeros((n,1)), np.ones((n,1))))

print("X shape:", X.shape) # (400, 2)
print("y shape:", y.shape) # (400, 1)   
print("First 5 samples:\n", X[:5])
print("First 5 labels:\n", y[:5])
```

```python
# 📌 np.random.rand() → Uniform [0,1)

import numpy as np

x = np.random.rand(5)        # 1D
x2 = np.random.rand(3,2)     # 2D
print("1D array:\n", x)
print("2D array:\n", x2)
```

```python
# 📌 np.random.uniform(low, high, size)

x = np.random.uniform(5, 10, 5)
x2 = np.random.uniform(-2, 2, (3,2))
print("Uniform [5,10):\n", x)
print("Uniform [-2,2):\n", x2)
```

```python
# 🎯 2️⃣ Normal (Gaussian) Distribution
# 📌 np.random.randn() → Mean=0, Std=1

x = np.random.randn(5)
x2 = np.random.randn(3,2)
print("Standard Normal (1D):\n", x)
print("Standard Normal (2D):\n", x2)
```

```python
# 📌 np.random.normal(mean, std, size)

x = np.random.normal(10, 2, 5)
x2 = np.random.normal(0, 1, (4,3))
print("Normal (mean=10, std=2):\n", x)
print("Standard Normal (mean=0, std=1):\n", x2)
```

```python
# 🎯 4️⃣ Random Choice
# 📌 np.random.choice()

# From list
x1 = np.random.choice([0,1], 10)

# With probabilities
x2 = np.random.choice([0,1], 10, p=[0.8,0.2])

# Without replacement
x3 = np.random.choice(10, 5, replace=False)

print("Random choice from [0,1]:\n", x1)
print("Random choice with probabilities:\n", x2)    
print("Random choice without replacement:\n", x3)
```

```python
# 🎯 5️⃣ Shuffle
# 📌 np.random.shuffle()

x = np.arange(10)
np.random.shuffle(x)
print("Shuffled array:\n", x)
```

```python
# 📌 np.random.permutation()

x = np.random.permutation(10)
print("Shuffled array (permutation):\n", x)
```

```python
# 🎯 6️⃣ Multivariate Normal (Advanced Gaussian)

mean = [0,0]
cov = [[1,0.8],[0.8,1]]

x = np.random.multivariate_normal(mean, cov, 200)
print("Multivariate Normal samples:\n", x[:5])
```

```python
# 🎯 7️⃣ Binomial Distribution
x1 = np.random.binomial(n=10, p=0.5, size=5)

# 🎯 8️⃣ Bernoulli Distribution
x2 = np.random.binomial(1, 0.7, 10)

print("Binomial samples:\n", x1)
print("Bernoulli samples:\n", x2)
```

```python
# 🎯 9️⃣ Exponential Distribution
x1 = np.random.exponential(scale=1.0, size=5)

# 🎯 🔟 Poisson Distribution
x2 = np.random.poisson(lam=3, size=5)

# 🎯 1️⃣1️⃣ Beta Distribution
x3  = np.random.beta(a=2, b=5, size=5)


# 🎯 1️⃣2️⃣ Gamma Distribution
x4 = np.random.gamma(shape=2, scale=2, size=5)

# 🎯 1️⃣3️⃣ Lognormal Distribution
x5 = np.random.lognormal(mean=0, sigma=1, size=5)

print("Exponential samples:\n", x1)
print("Poisson samples:\n", x2)
print("Beta samples:\n", x3)
print("Gamma samples:\n", x4)
print("Lognormal samples:\n", x5)
```
