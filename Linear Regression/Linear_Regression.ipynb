```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing

# Load dataset
housing = fetch_california_housing()

# Use only one feature (Average number of rooms)
X = housing.data[:, 2]   # AveRooms column
y = housing.target

# Normalize X (important for gradient descent)
X = (X - X.mean()) / X.std()

# Convert to numpy
X = np.array(X)
y = np.array(y)
```

```python
m = 0
b = 0

learning_rate = 0.01
epochs = 1000
n = len(X)
```

```python
for i in range(epochs):
    
    y_pred = m*X + b
    
    dm = (-2/n) * np.sum(X * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)
    
    m = m - learning_rate * dm
    b = b - learning_rate * db
```

```python
print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
plt.scatter(X, y)
plt.plot(X, m*X + b)
plt.title("Linear Regression using Gradient Descent")
plt.show()
```
