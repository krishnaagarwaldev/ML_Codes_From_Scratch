```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Number of samples
n = 100

# Generate features (normal distribution)
x1 = np.random.normal(5, 2, n)
x2 = np.random.normal(10, 3, n)

# True weights
true_w0 = 4     # bias
true_w1 = 3
true_w2 = 2

# Generate target with noise
y = true_w0 + true_w1*x1 + true_w2*x2 + np.random.normal(0, 2, n)
```

```python
# Add column of 1s for bias
X = np.column_stack((np.ones(n), x1, x2))

# Initialize weights randomly
W = np.zeros(3)

learning_rate = 0.001
epochs = 1000000
```

```python
for i in range(epochs):
    y_pred = X.dot(W)
    error = y - y_pred
    gradient = (-2/n) * X.T.dot(error)
    W = W - learning_rate * gradient
```

```python
print("Learned Weights:", W)
```
