```python
import pandas as pd
import numpy as np

# Small Titanic-like dataset
data = {
    'Age': [22, 38, 26, 35, 28, 2, 40, 19],
    'Fare': [7.25, 71.28, 7.92, 53.10, 8.05, 21.07, 31.00, 7.85],
    'Sex': ['male','female','female','female','male','male','male','female'],
    'Pclass': [3,1,3,1,3,3,2,3],
    'Survived': [0,1,1,1,0,0,1,1]
}

df = pd.DataFrame(data)
```

```python
# Separate features
numerical_cols = ['Age', 'Fare']
categorical_cols = ['Sex', 'Pclass']

X = df.drop('Survived', axis=1)
y = df['Survived']

classes = y.unique()
print("Classes:", classes)

# Test sample
test_sample = {
    'Age': 30,
    'Fare': 20,
    'Sex': 'female',
    'Pclass': 2
}
```

```python
final_probs = {}

for c in classes:
    
    # Prior
    df_c = df[df['Survived'] == c]
    prior = len(df_c) / len(df)
    print(df_c)
    print(f"\nClass {c} Prior: {prior:.4f}")
    
    total_prob = prior
    
    # -------- Numerical (Gaussian) --------
    for col in numerical_cols:
        mean = df_c[col].mean()
        var = df_c[col].var()
        x = test_sample[col]
        print(col, "Mean:", mean, "Variance:", var)
        
        # Gaussian formula
        prob = (1 / np.sqrt(2 * np.pi * var)) * \
               np.exp(-((x - mean)**2) / (2 * var))
        
        total_prob *= prob
    
    # -------- Categorical --------
    for col in categorical_cols:
        count = np.sum(df_c[col] == test_sample[col])
        prob = count / len(df_c)
        total_prob *= prob
        print(col, "Count:", count, "Total in Class:", len(df_c), "Prob:", prob)
    
    final_probs[c] = total_prob

# Prediction
prediction = max(final_probs, key=final_probs.get)

print("Class Probabilities:", final_probs)
print("Predicted Class:", prediction)
```
